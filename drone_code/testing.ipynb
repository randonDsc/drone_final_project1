```python
from playwright.async_api import async_playwright

URL = "https://made-to-give.lovable.app/"
NUM_VISITS = 2
VIEW_TIME_MS = 3000

async def test_visits():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)

        try:
            for visit in range(1, NUM_VISITS + 1):
                context = await browser.new_context()

                try:
                    page = await context.new_page()
                    await page.goto(URL, wait_until="load")
                    await page.wait_for_timeout(VIEW_TIME_MS)

                    await page.get_by_role(
                        "link",
                        name="Browse one-off cake sessions",
                        exact=True,
                    ).click()

                    await page.wait_for_url("**/sessions?cta=primary")
                    await page.wait_for_timeout(VIEW_TIME_MS)

                    print(
                        f"Session {visit}/{NUM_VISITS}: "
                        f"reached {page.url}"
                    )

                except Exception as error:
                    print(f"Session {visit}/{NUM_VISITS} failed: {error}")

                finally:
                    await context.close()

        finally:
            await browser.close()

# In Jupyter, use await directly.
await test_visits()
```
